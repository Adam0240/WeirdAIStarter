## Section 2: Layer Normalization
Implement normalization manually

```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

x = torch.tensor([
    [1.0, 2.0, 3.0],
    [5.0, 10.0, 15.0]
])

# Calculate - See pages 100 - 104
mean = x.mean(dim=-1, keepdim=True)
variance = x.var(dim=-1, keepdim=True, unbiased=False)

# Normalize each input using its mean and variance
normalized_output = (x - mean) / torch.sqrt(variance)

print("Mean:\n", mean)
print("Variance:\n", variance)
print("Normalized Output:\n", normalized_output)
```

```python
# Verify with
normalized_output.mean(dim=-1)
normalized_output.var(dim=-1, unbiased=False)
```

## Section 3: Exploring GELU
Compare ReLU and GELU

```python
# Creates 100 values between -4 and 4
x = torch.linspace(-4, 4, 100)

relu = nn.ReLU()
gelu = nn.GELU()

# Plot ReLU and GELU for comparison
plt.plot(x, relu(x), label="ReLU")
plt.plot(x, gelu(x), label="GELU")

plt.legend()
plt.show()
```

## Section 4: Feed Forward Network
Review figure 4.10 in the book

```python
# Create
input_shape = (2, 3, 768)

x = torch.rand(input_shape)

linear1 = nn.Linear(768, 3072)
gelu = nn.GELU()
linear2 = nn.Linear(3072, 768)

print("Input Shape:", x.shape)
```

```python
# Pass through 
# Linear
x = linear1(x)
print("After Linear:", x.shape)

# GELU
x = gelu(x)
print("After GELU:", x.shape)

# Linear again
x = linear2(x)
print("After Second Linear:", x.shape)
```

## Section 5: Residual Connections
Inspect output shapes using a toy network
Compare
 - With residual
 - Without residual

```python
# Creates a layer that keeps the same dimensions
layer = nn.Linear(768, 768)

# Without residual connection
without_residual = layer(x)

# With residual connection
output = x + layer(x)

print("Input Shape:", x.shape)
print("Without Residual:", without_residual.shape)
print("With Residual:", output.shape)
```
